```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv, find_dotenv
import os 

load_dotenv(find_dotenv(),override=True)

if os.environ['GOOGLE_API_KEY']:
    print("API Key found Sucess")
else:
    print("API Key not found please provide one")
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

# **Graph Schema**

```python
from typing import TypedDict,List,Annotated
from langgraph.graph.message import add_messages

class graph_schema(TypedDict):
    messages=Annotated[List,add_messages]
```

# **Node Functions**

```python
def welcome(state:graph_schema)->graph_schema:
    curr_message=state['messages']
    response=llm.invoke(curr_message).content
    state['messages']=f'Your Mesage was : {curr_message} \n Here is my response :{response}'

    return state
```

# **State Graph**

```python
from langgraph.graph import StateGraph,START,END

graph=StateGraph(state_schema=graph_schema)

graph.add_node("welcome",welcome)

graph.add_edge(START,"welcome")
graph.add_edge("welcome",END)
```
